# Face Emotion Recognition Demo
This notebook demonstrates face emotion recognition using OpenCV and DeepFace on sample images.

## 1. Setup & Imports
Import required libraries: OpenCV for image loading/processing, Matplotlib for displaying figures, and DeepFace for AI model analysis.

```python
import cv2
import matplotlib.pyplot as plt
from deepface import DeepFace
%matplotlib inline
```

## 2. Load Sample Images

```python
# Load sample test images
img1 = cv2.imread("samples/happy_boy.jpg")
img2 = cv2.imread("samples/sad_woman.jpg")

# Display Happy Boy sample (converting BGR to RGB for matplotlib display)
plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(img1, cv2.COLOR_BGR2RGB))
plt.title("Sample Image: Happy Boy")
plt.axis("off")
plt.show()
```

## 3. Analyze Emotion with DeepFace

```python
# Analyze emotions for sample image 1
analysis1 = DeepFace.analyze(img1, actions=["emotion"], enforce_detection=False, silent=True)
if isinstance(analysis1, list):
    analysis1 = analysis1[0]

print("Dominant Emotion:", analysis1["dominant_emotion"].capitalize())
print("Confidence:", f"{analysis1['emotion'][analysis1['dominant_emotion']]:.2f}%")
```

## 4. Draw Face Bounding Box & Annotations

```python
# Draw detection rectangle and emotion text overlay
img1_annotated = img1.copy()
region = analysis1["region"]
x, y, w, h = region["x"], region["y"], region["w"], region["h"]

cv2.rectangle(img1_annotated, (x, y), (x + w, y + h), (0, 255, 0), 2)
emotion_label = f"{analysis1['dominant_emotion'].capitalize()}"
cv2.putText(img1_annotated, emotion_label, (x, max(y - 10, 25)), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)

plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(img1_annotated, cv2.COLOR_BGR2RGB))
plt.title("Annotated Emotion Output")
plt.axis("off")
plt.show()
```

## 5. Analyze Second Sample Image (Sad Woman)

```python
analysis2 = DeepFace.analyze(img2, actions=["emotion"], enforce_detection=False, silent=True)
if isinstance(analysis2, list):
    analysis2 = analysis2[0]

print("Dominant Emotion:", analysis2["dominant_emotion"].capitalize())
print("Confidence:", f"{analysis2['emotion'][analysis2['dominant_emotion']]:.2f}%")

img2_annotated = img2.copy()
region2 = analysis2["region"]
x2, y2, w2, h2 = region2["x"], region2["y"], region2["w"], region2["h"]
cv2.rectangle(img2_annotated, (x2, y2), (x2 + w2, y2 + h2), (0, 255, 0), 2)
cv2.putText(img2_annotated, f"{analysis2['dominant_emotion'].capitalize()}", (x2, max(y2 - 10, 25)), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)

plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(img2_annotated, cv2.COLOR_BGR2RGB))
plt.title("Annotated Emotion Output: Sad Woman")
plt.axis("off")
plt.show()
```
